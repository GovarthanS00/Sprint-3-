# 1. lower()

## Definition
`lower()` converts all characters in a string to lowercase.

## Explanation
The original string contains uppercase letters.
`lower()` converts them all to lowercase.

## Business Example
Converting customer names or email addresses to lowercase for consistent data storage.

## AI/ML Example
Text preprocessing often converts text to lowercase so that
"Python", "PYTHON", and "python" are treated as the same word.

```python
## Python Example

text = "Hello PYTHON"

result = text.lower()

print(result)
```

```python
# 2. upper()

## Definition
`upper()` converts all characters in a string to uppercase.


## Explanation
The original string contains lowercase letters.
`upper()` converts all letters to uppercase.

## Business Example
Useful for standardizing text such as country codes, department names, or IDs.

## AI/ML Example
Can be used during text preprocessing when you want all text in a consistent uppercase format.
```

```python

## Python Example

text = "Hello Python"

result = text.upper()

print(result)
```

# 3. strip()

## Definition
`strip()` removes extra spaces from the beginning and end of a string.

## Explanation
The original string has spaces before and after the text.
`strip()` removes those unnecessary spaces.

## Business Example
Cleaning customer names or other text entered with unwanted spaces.

## AI/ML Example
Useful during data preprocessing to clean text before training a model.

```python

## Python Example

text = "   Hello Python   "

result = text.strip()

print(result)
```

# 4. split()

## Definition
`split()` breaks a string into multiple parts and returns them as a list.

## Explanation
`split()` separates the string wherever it finds a space.

"Python is easy"
        ↓
['Python', 'is', 'easy']

## Business Example
Splitting a full name, address, or sentence into separate parts.

## AI/ML Example
Useful in Natural Language Processing (NLP) to break sentences into individual words (tokens).

Example:
"Machine learning is useful"
        ↓
['Machine', 'learning', 'is', 'useful']

```python
## Python Example

text = "Python is easy"

result = text.split()

print(result)
```

# 5. contains()

## Definition
`contains()` checks whether a string contains a particular word or pattern.

## Explanation
`contains("Python")` checks each value:

"Python is easy"       → True
"Java is popular"      → False
"Python is powerful"   → True

## Business Example
Finding customer records that contain a specific word.

## AI/ML Example
Useful for filtering text data based on keywords.

Example:
Find reviews containing "good", "bad", or "excellent".

```python

## Python Example

import pandas as pd

data = pd.Series(["Python is easy", "Java is popular", "Python is powerful"])

result = data.str.contains("Python")

print(result)
```

# 6. replace()

## Definition
`replace()` replaces one string or value with another string or value.

## Explanation
Original:
"I like Java"

We replace `"Java"` with `"Python"`.

Result:
"I like Python"

## Business Example
Correcting or standardizing text values in a dataset.

Example:
"Chennai", "CHENNAI", "chennai"
can be standardized to:
"Chennai"

## AI/ML Example
Useful in text preprocessing to remove or replace unwanted words, symbols, or characters before training a model.

```python

## Python Example

text = "I like Java"

result = text.replace("Java", "Python")

print(result)
```

# 7. Regular Expressions (Regex)

## Definition
Regular Expression (Regex) is a pattern used to find, extract, or replace specific text.

Python uses the `re` module for Regex.

## Explanation
`\d+` means:
- `\d` → find a digit (0–9)
- `+` → find one or more digits

So Regex finds `9876543210` from the sentence.

## Business Example
Finding phone numbers, email addresses, or order IDs from customer data.

## AI/ML Example
Useful in text preprocessing to extract or remove specific patterns such as:
- Numbers
- Email addresses
- URLs
- Special characters
- Dates

Example:
"Email: user@gmail.com"
        ↓
Regex can extract:
"user@gmail.com"

```python

## Python Example

import re

text = "My phone number is 9876543210"

result = re.findall(r"\d+", text)

print(result)
```
