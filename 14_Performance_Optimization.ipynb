## 1. Vectorization

Vectorization applies operations to an entire Series or array without writing an explicit Python loop. It is often faster and clearer.

### Example

Calculate total sales using a vectorized operation.

```python
import pandas as pd

df = pd.DataFrame({
    "Price": [100, 200, 150],
    "Quantity": [2, 3, 4]
})
df["Total"] = df["Price"] * df["Quantity"]
print(df)
```

### Output Explanation

Pandas calculates all row-wise products in one vectorized expression.

## 2. `apply()` vs Loops

`apply()` applies a function along a Series or DataFrame axis. Vectorized expressions are usually preferable when available; `apply()` is not automatically faster than loops.

### Example

Apply a simple transformation to a column.

```python
df = pd.DataFrame({"Score": [40, 65, 85]})
df["Result"] = df["Score"].apply(lambda score: "Pass" if score >= 50 else "Fail")
print(df)
```

### Output Explanation

The function labels each score as Pass or Fail. For this simple condition, vectorized alternatives may be more efficient on large datasets.

## 3. Memory Optimization

Memory optimization reduces the memory used by a DataFrame, for example by choosing suitable numeric types and converting repeated strings to category.

### Example

Inspect memory usage before and after converting a repeated text column to category.

```python
df = pd.DataFrame({
    "Department": ["Sales", "IT", "Sales", "IT", "Sales"],
    "Employees": [10, 20, 15, 25, 12]
})
before = df.memory_usage(deep=True).sum()
df["Department"] = df["Department"].astype("category")
after = df.memory_usage(deep=True).sum()

print("Before:", before, "bytes")
print("After:", after, "bytes")
```

### Output Explanation

The printed values show the DataFrame memory usage before and after the conversion. The result depends on the data and category cardinality.

## 4. Category dtype

The `category` dtype stores repeated labels as categorical values and can reduce memory use when a column has relatively few unique values.

### Example

Convert a department column to category and inspect its type.

```python
df = pd.DataFrame({"Department": ["HR", "IT", "HR", "Sales", "IT"]})
df["Department"] = df["Department"].astype("category")
print(df.dtypes)
print(df["Department"].cat.categories)
```

### Output Explanation

The Department column has category dtype, and its unique categories are listed.

## 5. Efficient Pandas Practices

Efficient workflows avoid unnecessary copies, read only required columns when possible, use suitable data types, and measure performance instead of assuming.

### Example

Read selected columns from a CSV and inspect only a few rows. This example uses a small in-memory DataFrame to run without an external file.

```python
df = pd.DataFrame({
    "Name": ["Tony", "David", "John"],
    "Age": [25, 30, 28],
    "Salary": [35000, 42000, 39000]
})

selected = df[["Name", "Salary"]]
print(selected.head(2))
```

### Output Explanation

Selecting only required columns makes the next steps work with less data. For real files, `pd.read_csv()` can use `usecols` to read only needed columns.

# Best Practices

- Prefer vectorized operations where practical.
- Read only the columns and rows needed.
- Choose data types based on value ranges and meaning.
- Consider `category` for repeated labels with relatively few unique values.
- Measure runtime and memory before and after changes.

# Common Mistakes

- Assuming `apply()` is always faster than a loop or vectorized expression.
- Converting every text column to `category` without checking cardinality.
- Optimizing code before measuring its actual bottleneck.
- Making unnecessary DataFrame copies.
- Using numeric types that cannot safely hold the values.

# Advantages

- Can reduce memory usage.
- Can improve processing speed.
- Makes data workflows more scalable.
- Helps prepare larger datasets for analysis and modeling.

# Limitations

- The best optimization depends on the dataset and operation.
- Smaller data types can cause overflow if chosen incorrectly.
- Category dtype may not help columns with many unique values.
- Some optimizations can make code less readable if overused.
