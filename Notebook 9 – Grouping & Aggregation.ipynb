# 1.groupby()

## Definition
`groupby()` is used to split data into groups based on one or more columns and perform calculations on each group.

## Explanation
Rows are grouped by **Department**, then the salaries in each group are added together.

## Business Example
Calculate the total sales made by each branch.

## AI/ML Example
Find the average target value for each category before feature engineering.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "HR", "IT", "Sales"],
    "Salary": [30000, 50000, 35000, 60000, 45000]
})

result = df.groupby("Department")["Salary"].sum()
print(result)
```

# 2.agg()

## Definition
`agg()` (Aggregate) applies one or more aggregate functions such as `sum`, `mean`, `min`, `max`, or `count` on grouped data.

## Explanation
Multiple calculations (sum, mean, max) are performed for each department.

## Business Example
Generate department-wise salary reports.

## AI/ML Example
Calculate multiple statistics for grouped features during exploratory data analysis.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "HR", "IT"],
    "Salary": [30000, 50000, 35000, 60000]
})

result = df.groupby("Department").agg({
    "Salary": ["sum", "mean", "max"]
})

print(result)
```

# 3.transform()

## Definition
`transform()` performs calculations on groups and returns a result with the same size as the original DataFrame.

## Explanation
Each row gets the average salary of its own department.

## Business Example
Compare each employee's salary with the department average.

## AI/ML Example
Create group-based features such as average purchase value per customer.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "HR", "IT"],
    "Salary": [30000, 50000, 35000, 60000]
})

df["Dept_Avg"] = df.groupby("Department")["Salary"].transform("mean")

print(df)

```

# 4.filter()

## Definition
`filter()` keeps only the groups that satisfy a given condition.

## Explanation
Only departments whose average salary is greater than **40,000** are kept.

## Business Example
Keep branches whose total sales exceed the target.

## AI/ML Example
Remove categories that have very few records before training a model.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "HR", "IT", "Sales"],
    "Salary": [30000, 50000, 35000, 60000, 45000]
})

result = df.groupby("Department").filter(
    lambda x: x["Salary"].mean() > 40000
)

print(result)
```

# 5.Pivot Tables

## Definition
A Pivot Table summarizes data by grouping rows and columns with an aggregation function.
## Explanation
The average salary is calculated for each combination of Department and Gender.

## Business Example
View total sales by Region and Product.

## AI/ML Example
Analyze feature distributions across multiple categories during EDA.

```python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "HR", "IT"],
    "Gender": ["M", "M", "F", "F"],
    "Salary": [30000, 50000, 35000, 60000]
})

result = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)

print(result)
   
```

# 6.Crosstab

## Definition
`pd.crosstab()` creates a frequency table showing how often combinations of categories occur.

## Explanation
Counts the number of Male and Female employees in each department.

## Business Example
Count customers by Region and Membership Type.

## AI/ML Example
Check class distribution across different categorical features before model training.

```python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "HR", "IT", "HR"],
    "Gender": ["M", "M", "F", "F", "F"]
})

result = pd.crosstab(
    df["Department"],
    df["Gender"]
)

print(result)
```
