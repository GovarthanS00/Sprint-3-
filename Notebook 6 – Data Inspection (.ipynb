## 1.head()
### Definition

head() returns the first 5 rows of a DataFrame by default. You can specify a number to display that many rows.


### Explanation
head() displays the first 5 rows by default.
It is used to quickly inspect the dataset after loading.
You can use head(n) to display the first n rows.
### Business Example

A retail company loads customer purchase data and uses head() to verify that customer IDs, product names, and purchase amounts are loaded correctly before generating sales reports.

### AI/ML Example

A data scientist loads a dataset using Pandas and calls head() to inspect the first few rows before cleaning the data, handling missing values, and training a machine learning model.

````python
### Python Example
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [25, 30, 35, 28, 40, 32],
    "Salary": [50000, 60000, 70000, 55000, 80000, 65000]
}

df = pd.DataFrame(data)

print(df.head())
````

## 2.tail()
### Definition

tail() returns the last 5 rows of a DataFrame by default. You can specify a number to display that many rows.


### Explanation
* tail() displays the last 5 rows by default.
* It is useful for checking the end of a dataset.
* You can use tail(n) to display the last n rows.
### Business Example

A bank checks the last few transaction records using tail() to ensure the latest transactions have been recorded correctly.

### AI/ML Example

After merging or appending datasets, a data scientist uses tail() to verify that the last rows were added correctly before proceeding with feature engineering or model training.

````python
### Python Example
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [25, 30, 35, 28, 40, 32],
    "Salary": [50000, 60000, 70000, 55000, 80000, 65000]
}

df = pd.DataFrame(data)

print(df.tail())
````

## 3.sample()
### Definition

sample() returns random row(s) from a DataFrame. It is useful for randomly inspecting data.

### Explanation
* sample() selects rows randomly from the DataFrame.
* sample(3) returns 3 random rows.
* It helps inspect different parts of the dataset instead of only the beginning or end.
### Business Example

An e-commerce company randomly selects a few customer orders using sample() to manually verify that order details are correct.

### AI/ML Example

A data scientist uses sample() to randomly inspect training data and ensure there are no missing values, incorrect labels, or duplicate records before training a machine learning mode

````python
## Python Example
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [25, 30, 35, 28, 40, 32],
    "Salary": [50000, 60000, 70000, 55000, 80000, 65000]
}

df = pd.DataFrame(data)

print(df.sample(3))
````

## 4. info()
### Definition

info() displays a summary of the DataFrame, including the number of rows, columns, data types, non-null values, and memory usage.

### Explanation
Shows the total number of rows and columns.
Displays the data type (int64, object, etc.) of each column.
Shows the number of non-null values.
Displays the memory used by the DataFrame.
### Business Example

An HR team checks employee data using info() to verify that all required fields (Name, Age, Salary, Department) are present and have the correct data types.

### AI/ML Example

Before preprocessing, a data scientist uses info() to identify missing values and verify feature data types so they can perform data cleaning, encoding, and feature engineering correctly.

````python
### Python Example
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "Salary": [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)

df.info()
````

# 5. `describe()`

## Definition

`describe()` generates descriptive statistics of numerical columns in a DataFrame.

---

## Explanation

- **count** → Number of non-null values.
- **mean** → Average value.
- **std** → Standard deviation.
- **min** → Smallest value.
- **25%** → First quartile.
- **50%** → Median.
- **75%** → Third quartile.
- **max** → Largest value.

---

## Business Example

A retail company uses `describe()` to analyze product prices and sales, helping identify the average price, highest sale, and lowest sale.

---

## AI/ML Example

Before training a machine learning model, a data scientist uses `describe()` to understand feature distributions, detect unusual values, and decide whether feature scaling or outlier treatment is needed.



````python
###```python
import pandas as pd

data = {
    "Age": [25, 30, 35, 28, 40],
    "Salary": [50000, 60000, 70000, 55000, 80000]
}

df = pd.DataFrame(data)

print(df.describe())
````

# 6. `shape`

## Definition

`shape` returns the number of **rows and columns** in a DataFrame as a tuple `(rows, columns)`.

---

## Explanation

- `4` → Total number of rows (records).
- `3` → Total number of columns (features).
- `shape` is an **attribute**, so **do not use parentheses** (`shape()` is incorrect).

---

## Business Example

An HR department checks the shape of the employee dataset to confirm it contains **4 employee records** and **3 employee details** before generating reports.

---

## AI/ML Example

Before training a machine learning model, a data scientist uses `shape` to verify the dataset size and ensure the expected number of samples and features are available.

````python
## Python Example

import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "Salary": [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)

print(df.shape)
````

# 8. `dtypes`

## Definition

`dtypes` returns the data type of each column in a DataFrame.

---

## Explanation

- **object** → Text/String values.
- **int64** → Integer values.
- **float64** → Decimal values.
- **bool** → True/False values.
- `dtypes` helps identify the data type of every column.
- `dtypes` is an **attribute**, so **do not use parentheses** (`dtypes()` is incorrect).

---

## Business Example

A banking company checks `dtypes` to ensure **Account_Number** is stored correctly, **Balance** is numeric, and **Customer_Name** is stored as text before generating reports.

---

## AI/ML Example

Before model training, a data scientist uses `dtypes` to identify categorical and numerical features so they can apply the correct preprocessing techniques, such as encoding categorical data and scaling numerical features.

````python
## Python Example


import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Salary": [50000.0, 60000.0, 70000.0],
    "Is_Employee": [True, True, False]
}

df = pd.DataFrame(data)

print(df.dtypes)
````

# 9. `memory_usage()`

## Definition

`memory_usage()` returns the amount of memory used by each column in a DataFrame.

---

## Explanation

- **Index** → Memory used by the DataFrame index.
- **Name** → Memory used by the `Name` column.
- **Age** → Memory used by the `Age` column.
- **Salary** → Memory used by the `Salary` column.
- Helps identify columns that consume more memory.

---

## Business Example

A company working with millions of customer records uses `memory_usage()` to identify memory-heavy columns and optimize storage, improving application performance.

---

## AI/ML Example

Large machine learning datasets can consume significant RAM. Data scientists use `memory_usage()` to optimize memory usage (e.g., converting `int64` to `int32` or `float64` to `float32`) so large datasets fit into memory and models train more efficiently.

````python

## Python Example


import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Salary": [50000, 60000, 70000]
}

df = pd.DataFrame(data)

print(df.memory_usage())


````

> **Note:** The exact memory values may vary depending on your system and Pandas version.

# 10. `nunique()`

## Definition

`nunique()` returns the number of **unique (distinct)** values in each column of a DataFrame.

---


## Explanation

- **Department** has **3 unique values**: HR, IT, Finance.
- **Gender** has **2 unique values**: Male, Female.
- `nunique()` counts only distinct values and ignores duplicates.

---

## Business Example

A company uses `nunique()` to find how many unique departments, job roles, or branch locations exist in its employee database.

---

## AI/ML Example

A data scientist uses `nunique()` to identify categorical features and understand how many unique values each feature contains. This helps decide which encoding technique (such as Label Encoding or One-Hot Encoding) should be used before training a machine learning model.

````python

## Python Example


import pandas as pd

data = {
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
}

df = pd.DataFrame(data)

print(df.nunique())
````

# 11. `value_counts()`

## Definition

`value_counts()` counts the frequency of each unique value in a column.

---

## Explanation

- **IT** appears **3 times**.
- **HR** appears **2 times**.
- **Finance** appears **1 time**.
- The values are displayed in **descending order of frequency** by default.

---

## Business Example

A retail company uses `value_counts()` to find how many orders were placed in each product category, helping identify the most popular products.

---

## AI/ML Example

A data scientist uses `value_counts()` to check the distribution of target classes (e.g., **Spam** vs **Not Spam**, **Fraud** vs **Not Fraud**). This helps identify **class imbalance**, which is important before training a machine learning model.

````python
## Python Example

import pandas as pd

data = {
    "Department": ["HR", "IT", "IT", "Finance", "HR", "IT"]
}

df = pd.DataFrame(data)

print(df["Department"].value_counts())
````
