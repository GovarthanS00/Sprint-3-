## 1.Pandas
### Definition

Pandas is a Python library used to load, clean, analyze, and manipulate structured data efficiently.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)

print(df)
```

### Explanation
* import pandas as pd → Imports the Pandas library.
* pd.DataFrame(data) → Creates a table from a dictionary.
* print(df) → Displays the DataFrame.
### Business Example

A company uses Pandas to analyze customer sales data and generate reports.

### AI/ML Example

In AI/ML, Pandas is used to load datasets, clean missing values, filter data, create features, and prepare data before training machine learning models.

## 2.Series
### Definition

A Series is a one-dimensional labeled array in Pandas. It stores a single column of data with an index.

```python
import pandas as pd

marks = pd.Series([85, 92, 78, 90])

print(marks)
```

### Explanation
* pd.Series() → Creates a Pandas Series.
* [85, 92, 78, 90] → Values stored in the Series.
* 0, 1, 2, 3 → Default index values.
* dtype: int64 → Data type of the Series.
### Business Example

A company stores the daily sales of a single product in a Series for quick analysis.

### AI/ML Example

In AI/ML, a Series is commonly used to store a single feature (column) or the target variable (label), such as house prices, customer churn, or exam scores.

## 3.DataFrame
### Definition

A DataFrame is a two-dimensional table in Pandas that stores data in rows and columns. It is the most commonly used data structure in Pandas.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 28, 23],
    "City": ["Chennai", "Bangalore", "Hyderabad"]
}

df = pd.DataFrame(data)

print(df)
```

### Explanation
* pd.DataFrame() → Creates a DataFrame.
* Dictionary keys (Name, Age, City) become column names.
* Each list becomes a column.
* Each row represents one record.

### Business Example

A company stores employee details (Employee ID, Name, Department, Salary) in a DataFrame for reporting and analysis.

### AI/ML Example

In AI/ML, a DataFrame is used to store the entire dataset, where each row is one data sample and each column is a feature (or the target column) before preprocessing and model training.

## Creating DataFrames
### Definition

A DataFrame can be created from different data sources such as dictionaries, lists, tuples, NumPy arrays, CSV files, Excel files, and SQL databases.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 28, 23],
    "City": ["Chennai", "Bangalore", "Hyderabad"]
}

df = pd.DataFrame(data)

print(df)
```

### Explanation
* Create a Python dictionary.
* Pass the dictionary to pd.DataFrame().
* Each dictionary key becomes a column.
* Each list becomes the column values.
* The DataFrame is created and displayed.

### Business Example

A company creates a DataFrame from employee records to analyze salaries, departments, and attendance.

### AI/ML Example

In AI/ML, DataFrames are created after loading datasets from CSV, Excel, SQL, or APIs. The DataFrame is then used for data cleaning, preprocessing, feature engineering, and model training.

## 4. Data Types
### Definition

Data types define the type of values stored in each DataFrame column, such as integers, floats, strings, or booleans.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 28, 23],
    "Salary": [45000.50, 52000.75, 40000.00],
    "Placed": [True, True, False]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

### Explanation
* object → Text (strings)
* int64 → Whole numbers
* float64 → Decimal numbers
* bool → True or False
* df.dtypes → Displays the data type of each column.

### Business Example

A bank stores customer names as text, account balances as decimal numbers, and loan status as True/False.

### AI/ML Example

In AI/ML, checking data types is important because machine learning models require numerical data. Text and categorical columns are usually converted into numeric values before training the model.

## 5.Index
### Definition

An Index is a unique label assigned to each row in a DataFrame or Series. By default, Pandas starts the index from 0.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 28, 23]
}

df = pd.DataFrame(data)

print(df.index)
```

## Explanation
* df.index → Displays the row index.
* start=0 → Index starts from 0.
* stop=3 → Stops before 3 (0, 1, 2).
* step=1 → Increments by 1.
### Business Example

A company uses the index to uniquely identify each sales record or transaction.

### AI/ML Example

In AI/ML, the index helps identify and track each data sample during preprocessing, filtering, and train-test splitting. The index itself is usually not used as a feature for model training.

## 6. Columns
### Definition

Columns are the vertical sections of a DataFrame. Each column represents a feature or attribute of the data.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 28, 23],
    "City": ["Chennai", "Bangalore", "Hyderabad"]
}

df = pd.DataFrame(data)

print(df.columns)
```

### Explanation
* df.columns → Displays all column names.
* Each column stores one type of information (feature).
* Column names are used to access and manipulate data.
### Business Example

A customer database may have columns like Customer_ID, Name, Age, and City.

### AI/ML Example

In AI/ML, each column represents a feature (input variable) used to train a machine learning model. The target column contains the value the model needs to predict.

## 7.Reading Data
### Definition

Reading Data means loading data from files or databases into a Pandas DataFrame for analysis.

### Explanation

* The data is loaded into a DataFrame.
* print(df) displays the loaded data.
### Business Example

A retail company reads daily sales data from a CSV file to generate sales reports.

### AI/ML Example

In AI/ML, datasets are usually loaded from CSV, Excel, SQL databases, or APIs into a DataFrame before data cleaning, preprocessing, feature engineering, and model training.

## 8.Writing Data
### Definition

Writing Data means saving a Pandas DataFrame to a file such as CSV, Excel, or JSON.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("Data saved successfully!")
```

### Explanation
* pd.DataFrame(data) → Creates a DataFrame.
* df.to_csv("students.csv", index=False) → Saves the DataFrame as a CSV file.
* index=False → Prevents the row index from being saved.
* print() → Displays a success message.
### Business Example

A company saves processed sales data as a CSV file to share with the finance team.

### AI/ML Example

In AI/ML, cleaned or transformed datasets are saved as CSV or Excel files so they can be reused for model training, testing, or deployment without repeating the preprocessing steps.
