## CSV Files
### Definition

A CSV (Comma-Separated Values) file stores tabular data in rows and columns separated by commas. It is the most commonly used format for datasets.

import pandas as pd

df = pd.read_csv("students.csv")

print(df)

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78 
```

### Explanation
* pd.read_csv() reads a CSV file.
* The data is loaded into a Pandas DataFrame.
* print(df) displays the dataset.

### Business Example

Companies store sales reports, customer records, and transaction data in CSV files because they are simple and widely supported.

### AI/ML Example

Most AI/ML datasets (such as Kaggle datasets) are provided in CSV format. We load them using Pandas before data cleaning, preprocessing, and model training.

## Excel Files
### Definition

An Excel file (.xlsx or .xls) stores data in one or more worksheets (sheets). Pandas can read and write Excel files.

```python
import pandas as pd

df = pd.read_excel("students.xlsx")

print(df)
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
```

### Explanation
* pd.read_excel() reads an Excel file.
* The data is loaded into a DataFrame.
* print(df) displays the dataset.

Note: Reading Excel files requires the openpyxl library

### Business Example

Companies maintain employee records, financial reports, and inventory data in Excel files for reporting and analysis.

### AI/ML Example

In AI/ML, datasets received from clients are often in Excel format. Pandas reads these files so the data can be cleaned, preprocessed, and used to train machine learning models.

## 3.JSON Files

```python
import pandas as pd

df = pd.read_json("students.json")

print(df)
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
```

## Parquet Files
### Definition

A Parquet file is a column-based storage format that is faster and more storage-efficient than CSV. It is commonly used for big data.

```python
import pandas as pd

df = pd.read_parquet("students.parquet")

print(df)
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
```

### Explanation
* pd.read_parquet() reads a Parquet file.
* The data is loaded into a DataFrame.
* print(df) displays the dataset.

Note: Reading Parquet files requires either pyarrow or fastparquet. - pip install pyarrow

### Business Example

Large organizations store millions of sales or transaction records in Parquet files because they require less storage and provide faster data access.

### AI/ML Example

In AI/ML, large datasets are often stored in Parquet format because it loads faster, uses less disk space, and improves preprocessing performance, especially when working with big data.

## 5.Pickle Files
### Definition

A Pickle file (.pkl) stores Python objects in a binary format. It is commonly used to save and load DataFrames or trained machine learning models

```python
import pandas as pd

df = pd.read_pickle("students.pkl")

print(df)  
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
```

### Explanation
* pd.read_pickle() reads a Pickle file.
* The stored DataFrame is loaded into memory.
* print(df) displays the data.
### Business Example

A company saves processed data as a Pickle file so employees can reload it quickly without repeating data processing.

### AI/ML Example

In AI/ML, Pickle files are used to save preprocessed DataFrames or trained machine learning models so they can be loaded later for prediction or deployment.

## 6.SQL Tables
### Definition

A SQL Table stores structured data inside a relational database. Pandas can read data from SQL tables and convert it into a DataFrame.

```python
import pandas as pd
import sqlite3

# Connect to the database
conn = sqlite3.connect("students.db")

# Read data from SQL table
df = pd.read_sql("SELECT * FROM students", conn)

print(df)

conn.close()
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
```

### Explanation
* sqlite3.connect() → Connects to the SQL database.
* pd.read_sql() → Reads data from the SQL table into a DataFrame.
* print(df) → Displays the data.
* conn.close() → Closes the database connection

### Business Example

Banks store customer accounts and transactions in SQL databases for secure storage and easy querying.

### AI/ML Example

In AI/ML, datasets are often stored in SQL databases. Pandas reads the required data into a DataFrame for preprocessing, feature engineering, and model training.

## 7.Reading Multiple Files
### Definition

Reading Multiple Files means loading data from multiple files and combining them into a single DataF

```python
import pandas as pd
import glob

# Read all CSV files
files = glob.glob("data/*.csv")

# Combine all files
df = pd.concat([pd.read_csv(file) for file in files], ignore_index=True)

print(df)
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
3   John   23     88
4   Alice  24     95
```

## Explanation
* glob.glob("data/*.csv") → Finds all CSV files in the data folder.
* pd.read_csv(file) → Reads each CSV file.
* pd.concat() → Combines all DataFrames into one.
* ignore_index=True → Creates a new continuous index.
### Business Example

A company combines monthly sales reports from different branches into a single dataset for yearly analysis.

### AI/ML Example

In AI/ML, datasets are often split into multiple files (daily, monthly, or by source). Pandas combines them into one DataFrame before preprocessing and model training.

## File Path Handling
### Definition

File Path Handling means specifying the correct location of a file so Pandas can read or write it.

```python
import pandas as pd

# Read a file using its path
df = pd.read_csv("data/students.csv")

print(df)
```

```python
    Name  Age  Marks
0   Ravi   21     85
1  Priya   22     92
2   Arun   20     78
```

### Explanation
* "data/students.csv" is the file path.
* pd.read_csv() uses the path to locate and read the file.
* If the path is incorrect, Pandas raises a FileNotFoundError.
### Business Example

A company stores reports in different folders (e.g., Sales/2026/July/). File paths help applications access the correct report automatically.

### AI/ML Example

In AI/ML, datasets, trained models, and prediction results are often stored in different folders. Proper file path handling ensures the correct files are loaded for training, testing, and deployment.
