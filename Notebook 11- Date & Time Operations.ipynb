## 1. Datetime

### Definition
`datetime` represents both **date and time** in Python.

It is useful for working with dates such as year, month, day, hour, minute, and second.

### Explanation
`datetime.now()` returns the current date and time.

For example:

2026 → Year
08 → Month
07 → Day
15 → Hour
30 → Minute
25 → Second

You can also create a specific datetime:

date = datetime(2026, 8, 7, 10, 30)

print(date)

### Output

2026-08-07 10:30:00

### Business Example
A company may store:

- Customer registration date
- Order date
- Payment date
- Login time
- Transaction time

Using `datetime`, these dates and times can be stored and analyzed.

### AI/ML Example
In machine learning, datetime data can be used to understand **when an event happened**.

For example:

Customer_ID | Login_Time
1           | 2026-08-01 09:30
2           | 2026-08-01 18:45

Datetime information can later be converted into useful features such as:

- Hour
- Day
- Month
- Day of week

These features can help models identify patterns such as customer activity by time.

````python

### Python Example

from datetime import datetime

now = datetime.now()

print(now)
````

## 2. Timestamp

### Definition
A `Timestamp` is a specific point in time used by Pandas to represent dates and times.

Pandas provides `pd.Timestamp()` for creating and working with timestamps.


### Explanation
`pd.Timestamp()` converts a date and time into a Pandas Timestamp object.

You can access individual parts:

print(timestamp.year)
print(timestamp.month)
print(timestamp.day)
print(timestamp.hour)

### Output

2026
8
7
10

You can also create the current timestamp:

print(pd.Timestamp.now())

### Business Example
A company can use timestamps to record:

- Order time
- Payment time
- Login time
- Transaction time
- System events

This helps analyze when business activities happened.

### AI/ML Example
Timestamp data can be converted into useful ML features.

For example:

timestamp = pd.Timestamp("2026-08-07 18:30:00")

hour = timestamp.hour

print(hour)

### Output

18

The `18` represents 6 PM.

This can help a model learn patterns such as:

- When customers usually purchase
- Peak website traffic hours
- Fraudulent transaction times
- User activity patterns

### Key Point

`datetime` → Python's date and time object.

`Timestamp` → Pandas' date and time object, especially useful when working with DataFrames and time-series data.

````python

### Python Example

import pandas as pd

timestamp = pd.Timestamp("2026-08-07")

print(timestamp)
````

## 3. Date Parsing

### Definition
Date parsing means converting date values stored as strings into a proper date/time format.

In Pandas, we use `pd.to_datetime()` for date parsing.

### Explanation
Initially, the dates are stored as strings:

"2026-08-01"

`pd.to_datetime()` converts them into Pandas datetime values.

After conversion, we can perform date operations such as:

* parsed_dates.dt.year
* parsed_dates.dt.month
* parsed_dates.dt.day

### Example with Different Date Format

dates = pd.Series([
    "07-08-2026",
    "08-08-2026",
    "09-08-2026"
])

parsed_dates = pd.to_datetime(
    dates,
    format="%d-%m-%Y"
)

print(parsed_dates)

### Output

* 0   2026-08-07
* 1   2026-08-08
* 2   2026-08-09
dtype: datetime64[ns]

### Business Example
A company may receive sales dates in different formats:

"07-08-2026"
"2026/08/07"
"August 7, 2026"

Date parsing converts these values into a proper datetime format so they can be analyzed consistently.

### AI/ML Example
Suppose a dataset contains:

Customer_ID    Purchase_Date
101            07-08-2026
102            15-08-2026
103            20-08-2026

Convert the column:

df["Purchase_Date"] = pd.to_datetime(
    df["Purchase_Date"],
    format="%d-%m-%Y"
)

Then extract useful features:

df["Year"] = df["Purchase_Date"].dt.year
df["Month"] = df["Purchase_Date"].dt.month
df["Day"] = df["Purchase_Date"].dt.day

These features can be used by ML models to identify time-based patterns.

### Key Point

String date
↓
pd.to_datetime()
↓
Datetime value

Date parsing is important because real-world datasets often store dates as strings.


````python

### Python Example

import pandas as pd

dates = pd.Series([
    "2026-08-01",
    "2026-08-02",
    "2026-08-03"
])

parsed_dates = pd.to_datetime(dates)

print(parsed_dates)
````

## 4. Timedelta

### Definition
`Timedelta` represents the **difference between two dates or times**.

It is used to calculate how many days, hours, minutes, etc. are between two dates.

### Explanation
The difference between:

2026-08-01 → Start date
2026-08-10 → End date

is:

9 days

So `Timedelta` tells us the **amount of time between two dates**.

### Another Example

start = pd.Timestamp("2026-08-07 10:00:00")
end = pd.Timestamp("2026-08-07 15:30:00")

difference = end - start

print(difference)

### Output

0 days 05:30:00

This means the difference is:

5 hours 30 minutes

### Business Example
A company wants to calculate how long it takes to deliver an order.

Order Date:
2026-08-01

Delivery Date:
2026-08-05

The Timedelta is:

4 days

This can be used to calculate average delivery time.

### AI/ML Example
In customer analytics, we may calculate:

Last Purchase Date
→
Current Date

The difference gives the **number of days since the customer's last purchase**.

For example:

```python
last_purchase = pd.Timestamp("2026-08-01")
today = pd.Timestamp("2026-08-10")

days_since_purchase = today - last_purchase

print(days_since_purchase.days)

````python
### Python Example

import pandas as pd

start_date = pd.Timestamp("2026-08-01")
end_date = pd.Timestamp("2026-08-10")

difference = end_date - start_date

print(difference)
````

## 5. Time Zones

### Definition
Time zone tells us the local time of a particular location.

Pandas provides `tz_localize()` and `tz_convert()` to work with time zones.

### Explanation
`tz_localize()` adds a time zone to a datetime.

Here:

"Asia/Kolkata" = Indian Standard Time (IST)

`+05:30` means India is 5 hours and 30 minutes ahead of UTC.

### Convert to Another Time Zone

date = pd.Timestamp("2026-08-07 10:00:00")

date = date.tz_localize("Asia/Kolkata")

converted = date.tz_convert("America/New_York")

print(converted)

### Output

2026-08-07 00:30:00-04:00

This means:

India → 10:00 AM
New York → 12:30 AM

Both represent the same actual moment, but the local times are different.

### Business Example
A global company may have employees and customers in different countries.

For example:

India → 10:00 AM
USA → 12:30 AM

Time zone conversion helps schedule meetings, process transactions, and analyze activities correctly.

### AI/ML Example
A global application may collect user activity from different countries.

For example:

User A → India → 10:00 AM
User B → USA → 10:00 AM

These are not necessarily the same actual time.

Converting timestamps to a common time zone such as UTC helps ML models correctly understand the timing and order of events.

Useful for:

- Time-series analysis
- User activity prediction
- Fraud detection
- Customer behavior analysis

### Key Point

`tz_localize()` → adds a time zone to a datetime.

`tz_convert()` → converts a timezone-aware datetime to another time zone.

Example:

India Time → UTC → USA Time

````python

### Python Example

import pandas as pd

date = pd.Timestamp("2026-08-07 10:00:00")

# Add a time zone
date = date.tz_localize("Asia/Kolkata")

print(date)
````

## 6. Extracting Date Features

### Definition
Extracting date features means getting useful information such as Year, Month, Day, Hour, and Day of Week from a date.

### Explanation

The `.dt` accessor is used to extract information from a datetime column.

Year:
2026

Month:
8

Day:
7

Hour:
10

Day_of_Week:
4

Day of week values:

0 = Monday
1 = Tuesday
2 = Wednesday
3 = Thursday
4 = Friday
5 = Saturday
6 = Sunday

So `4` means Friday.

### Business Example

An e-commerce company has an order date:

2026-08-07 10:30:00

The company can extract:

Year → 2026
Month → 8
Day → 7
Hour → 10
Day → Friday

This helps identify which days, months, or hours have more sales.

### AI/ML Example

Suppose we want to predict sales.

Original feature:

Purchase_Date = 2026-08-07 10:30:00

Extracted features:

Year = 2026
Month = 8
Day = 7
Hour = 10
Day_of_Week = 4

These values can be used as input features for a Machine Learning model.

For example, the model may learn:

- Weekend → higher sales
- Evening → higher purchases
- Certain months → higher demand

### Key Point

Extracting Date Features means:

Date/Time
↓
Year + Month + Day + Hour + Day of Week
↓
Useful ML Features

````python
### Python Example

import pandas as pd

df = pd.DataFrame({
    "Date": ["2026-08-07 10:30:00"]
})

df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Hour"] = df["Date"].dt.hour
df["Day_of_Week"] = df["Date"].dt.dayofweek

print(df)

````
