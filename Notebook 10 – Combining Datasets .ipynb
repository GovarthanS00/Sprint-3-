## 1. concat()

### Definition
`concat()` combines two or more Pandas DataFrames along rows or columns.

### Explanation
`concat()` simply stacks DataFrames together.

- `axis=0` → combine rows (default)
- `axis=1` → combine columns
- `ignore_index=True` → reset the index

Example:

result = pd.concat([df1, df2], ignore_index=True)

### Business Example
Combine monthly sales data:

January Sales + February Sales → One Sales DataFrame.

### AI/ML Example
Combine training data collected from different files or sources into one dataset before model training.

````python

### Python Example

import pandas as pd

df1 = pd.DataFrame({
    "Name": ["A", "B"],
    "Score": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["C", "D"],
    "Score": [70, 85]
})

result = pd.concat([df1, df2])

print(result)
````

## 2. merge()

### Definition
`merge()` combines two Pandas DataFrames using one or more common columns. It works similar to SQL JOIN.

### Explanation
`merge()` connects two DataFrames using a common column.

Here, `Customer_ID` exists in both DataFrames, so Pandas uses it to match the records.

`on="Customer_ID"` specifies the common column.

Common merge types:

- `inner` → only matching records
- `left` → all records from left DataFrame
- `right` → all records from right DataFrame
- `outer` → all records from both DataFrames

Example:

result = pd.merge(customers, orders, on="Customer_ID", how="left")

### Business Example
A company has:

Customer Data:
`Customer_ID, Name`

Order Data:
`Customer_ID, Amount`

Using `merge()`, we can combine them to find how much each customer purchased.

### AI/ML Example
Suppose we have:

Customer Data:
`Customer_ID, Age, City`

Transaction Data:
`Customer_ID, Amount, Purchase_Count`

We can use `merge()` to create one dataset containing all customer and transaction features.

This combined dataset can then be used for:

- Customer churn prediction
- Fraud detection
- Customer segmentation
- Recommendation systems

````python
### Python Example

import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

orders = pd.DataFrame({
    "Customer_ID": [1, 2, 3],
    "Amount": [500, 800, 300]
})

result = pd.merge(customers, orders, on="Customer_ID")

print(result)
````

## 3. join()

### Definition
`join()` combines two Pandas DataFrames using their indexes by default.

### Explanation
`join()` combines DataFrames based on their **index**.

Here:

- `df1` has index `1, 2, 3`
- `df2` has index `1, 2, 3`
- Pandas matches these indexes and combines the columns.

By default, `join()` performs a left join.

You can also specify:

```python
df1.join(df2, how="inner")
df1.join(df2, how="outer")
df1.join(df2, how="right")

````python

### Python Example

import pandas as pd

df1 = pd.DataFrame({
    "Name": ["A", "B", "C"]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "Score": [80, 90, 70]
}, index=[1, 2, 3])

result = df1.join(df2)

print(result)
````

## 4. append() — Deprecated

### Definition
`append()` was previously used to add rows from one Pandas DataFrame to another.

However, `DataFrame.append()` was **deprecated in Pandas 1.4** and **removed in Pandas 2.0**.

So, we should use `concat()` instead.

### Explanation
Earlier, we could use:

df1.append(df2)

to add rows.

But `append()` was removed from modern Pandas.

Use:

pd.concat([df1, df2], ignore_index=True)

instead.

`ignore_index=True` creates a new continuous index:

0, 1, 2

### Business Example
A company receives daily sales data.

Day 1:
`df1`

Day 2:
`df2`

Instead of using the old `append()` method, use `concat()` to combine the daily sales records.

### AI/ML Example
Suppose ML training data is collected from multiple batches:

Batch 1 → `df1`

Batch 2 → `df2`

Batch 3 → `df3`

Combine them using:

training_data = pd.concat(
    [df1, df2, df3],
    ignore_index=True
)

This creates one training dataset for model training.

### Key Point

❌ Do not use:

df1.append(df2)

✅ Use:

pd.concat([df1, df2], ignore_index=True)

`concat()` is the modern replacement for `append()`.

````python

### Old Python Example

import pandas as pd

df1 = pd.DataFrame({
    "Name": ["A", "B"],
    "Score": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["C"],
    "Score": [70]
})

# Old method — no longer recommended
# result = df1.append(df2)

### Alternative using concat()

result = pd.concat([df1, df2], ignore_index=True)

print(result)


````

## 5. combine_first()

### Definition
`combine_first()` fills missing values in one DataFrame using values from another DataFrame.

It keeps the values from the first DataFrame when they are available and uses the second DataFrame only where values are missing.

### Explanation
`combine_first()` checks the first DataFrame:

- If a value exists → keep it.
- If the value is missing (`NaN`) → take the value from the second DataFrame.

For example:

df1 Score:
80, NaN, 70

df2 Score:
85, 90, 75

Result:
80, 90, 70

The missing value `NaN` in `df1` is replaced with `90` from `df2`.

### Business Example
A company has customer information from two systems.

System 1 contains some missing customer values.

System 2 contains backup information.

`combine_first()` can fill the missing values from System 2 without replacing the values that already exist in System 1.

### AI/ML Example
Suppose a machine learning dataset has missing feature values:

Dataset 1:
`Age, Income, Score`

Some values are missing.

Dataset 2 contains backup values for those features.

`combine_first()` can fill the missing values before model training.

This can help when combining data from different sources or backup datasets.

### Key Point

`combine_first()` is mainly used for:

- Filling missing values
- Combining backup data
- Data integration
- Handling incomplete datasets

Example:

df1.combine_first(df2)

It means:

"Use df1 first, and use df2 only where df1 has missing values."

````python

### Python Example

import pandas as pd
import numpy as np

df1 = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Score": [80, np.nan, 70]
})

df2 = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Score": [85, 90, 75]
})

result = df1.combine_first(df2)

print(result)
````
