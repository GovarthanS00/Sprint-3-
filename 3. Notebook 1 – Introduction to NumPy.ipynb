## 13. COPY VS VIEW

### Definition:

### Copy:
Creates a completely independent array. Changes to the original do not affect the copy.

### View:
Creates another way to access the same underlying data. Changes can affect the original.

### AI/ML Use:
Important when modifying datasets because accidental changes to the original data can affect model training.

Example:

```python
import numpy as np

arr = np.array([1, 2, 3])

copy_arr = arr.copy()
view_arr = arr.view()

arr[0] = 100

print("Original:", arr)
print("Copy:", copy_arr)
print("View:", view_arr)
```

## 15.ARRAY OPERATIONS

### Definition:
NumPy allows mathematical operations to be performed directly on arrays.

### Common operations:

### AI/ML Use:

## 16.UNIVERSAL FUNCTIONS (ufuncs)

### Definition:
ufuncs are NumPy functions that perform fast element-wise operations on arrays.

### Common ufuncs:
np.sqrt()
np.exp()
np.log()
np.add()
np.subtract()
np.multiply()
np.divide()

### AI/ML Use:
Used for fast mathematical transformations and numerical calculations during data preprocessing and ML computations.

Example:

```python
import numpy as np

arr = np.array([1, 4, 9, 16])

print(np.sqrt(arr))
print(np.sum(arr))
print(np.mean(arr))
```
