# 1.Rename Columns

## Definition
`rename()` is used to change one or more column names without modifying the original DataFrame unless `inplace=True` is used.

## Explanation
The column **Emp_Name** is renamed to **Employee_Name**.

## Business Example
Rename `Cust_ID` to `Customer_ID` to make reports easier to understand.

## AI/ML Example
Rename `target` to `Price` or `label` to make datasets more readable before model training.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Emp_ID": [101, 102],
    "Emp_Name": ["John", "Alice"]
})

df = df.rename(columns={"Emp_Name": "Employee_Name"})
print(df)

```

# 2.Drop Rows

## Definition
`drop()` removes one or more rows using their index.

## Explanation
The row with index **1** (Alice) is removed.

## Business Example
Remove cancelled orders from the sales dataset.

## AI/ML Example
Drop corrupted or invalid records before training a machine learning model.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, 30, 28]
})

df = df.drop(1)
print(df)

```

# 3.Drop Columns

## Definition
`drop()` can also remove unnecessary columns from a DataFrame.

## Explanation
The **City** column is removed from the DataFrame.

## Business Example
Remove internal tracking columns before sharing customer data.

## AI/ML Example
Drop ID columns since they usually don't help machine learning models.

```python
 
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice"],
    "Age": [25, 30],
    "City": ["Chennai", "Bangalore"]
})

df = df.drop(columns=["City"])
print(df)

```

# 4.Insert Columns

## Definition
`insert()` adds a new column at a specific position.


## Explanation
The **Gender** column is inserted at position **1**.

## Business Example
Insert a new `Department` column into employee records.

## AI/ML Example
Insert a new feature column before training a model.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice"],
    "Age": [25, 30]
})

df.insert(1, "Gender", ["Male", "Female"])
print(df)

```

# 5.assign()

## Definition
`assign()` creates one or more new columns and returns a new DataFrame.

## Explanation
A new column **Tax** is created using the existing **Price** column.

## Business Example
Create a GST or Discount column from product prices.

## AI/ML Example
Generate new features (feature engineering) from existing columns.

```python

## Python Example
import pandas as pd

df = pd.DataFrame({
    "Price": [100, 200, 300]
})

df = df.assign(Tax=df["Price"] * 0.18)
print(df)

```

# 6.apply()

## Definition
`apply()` applies a function to each value, row, or column.


## Explanation
The function checks each mark and returns **Pass** or **Fail**.

## Business Example
Assign customer categories based on purchase amount.

## AI/ML Example
Convert numeric values into categorical labels during preprocessing.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Marks": [34, 85, 90]
})

df["Result"] = df["Marks"].apply(lambda x: "Pass" if x >= 35 else "Fail")
print(df)
```

# 7. map()

## Definition
`map()` replaces or transforms values in a Series.

## Explanation
Each value is replaced using the mapping dictionary.

## Business Example
Convert product status codes into readable names.

## AI/ML Example
Convert categorical labels into meaningful text before analysis.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Gender": ["M", "F", "M"]
})

df["Gender"] = df["Gender"].map({
    "M": "Male",
    "F": "Female"
})

print(df)

```

# 8.replace()

## Definition
`replace()` replaces one or more values in a DataFrame or Series.

## Explanation
The values **Y** and **N** are replaced with **Yes** and **No**.

## Business Example
Replace outdated department names with new ones.

## AI/ML Example
Replace missing category labels like `"Unknown"` with `"Others"` before training.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Status": ["Y", "N", "Y"]
})

df["Status"] = df["Status"].replace({
    "Y": "Yes",
    "N": "No"
})

print(df)
```

# 9.Duplicate Handling

## Definition
`duplicated()` identifies duplicate rows, and `drop_duplicates()` removes them.

## Explanation
`duplicated()` marks repeated rows, while `drop_duplicates()` keeps only unique records.

## Business Example
Remove duplicate customer records before generating reports.

## AI/ML Example
Remove duplicate training samples to improve data quality and prevent model bias.

```python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "John"],
    "Age": [25, 30, 25]
})

print(df.duplicated())

df = df.drop_duplicates()
print(df)
```
