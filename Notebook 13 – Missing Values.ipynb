## 1. isnull()

### Definition
`isnull()` checks whether values in a DataFrame or Series are missing (`NaN`).

### Explanation
- `False` → value is present.
- `True` → value is missing (`NaN`).
- Here, Alice's Age is missing.

### Business Example
Find missing customer information before using the data for analysis.

### AI/ML Example
Check for missing values before training a machine learning model, because many ML algorithms cannot directly handle `NaN` values.

```python

### Python Example
import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, None, 30]
})

print(df.isnull())
```

## 2. notnull()

### Definition
`notnull()` checks whether values are **present** (not missing).

### Explanation
- `True` → value is present.
- `False` → value is missing (`NaN`).
- Alice's Age is missing, so it returns `False`.

### Business Example
Find customers whose required information is available.

### AI/ML Example
Use `notnull()` to identify rows where important ML features are available before model training.

```python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, None, 30]
})

print(df.notnull())
```

## 3. dropna()

### Definition
`dropna()` removes rows or columns that contain missing values (`NaN`).

### Explanation
- Alice's row contains a missing Age.
- `dropna()` removes that entire row.
- John and Bob remain because they have no missing values.

### Business Example
Remove customer records that have incomplete information.

### AI/ML Example
Remove rows with missing values when only a small number of records are affected and deleting them will not significantly reduce the dataset.

```python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, None, 30]
})

df = df.dropna()

print(df)
```

## 4. fillna()

### Definition
`fillna()` replaces missing values (`NaN`) with a specified value.

### Explanation
- Alice's Age was missing (`NaN`).
- `fillna(0)` replaces the missing value with `0`.
- No rows are removed.

### Business Example
If a customer's missing purchase amount means no purchase was recorded, we may replace it with `0`.

### AI/ML Example
Replace missing numerical feature values with a suitable value such as the mean, median, or `0` before training a model.

```python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, None, 30]
})

df["Age"] = df["Age"].fillna(0)

print(df)
```

## 5. interpolate()

### Definition
`interpolate()` fills missing values by estimating a value between existing values.
### Explanation
The missing value is between `120` and `160`.

Python estimates:

(120 + 160) / 2 = 140

So the missing value becomes `140`.

### Business Example
If sales data is missing for one day, interpolation can estimate the sales based on nearby days.

### AI/ML Example
Useful for time-series data where missing values occur between known values, such as temperature, stock prices, or sensor readings.

```python
### Python Example

import pandas as pd

df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [100, 120, None, 160, 180]
})

df["Sales"] = df["Sales"].interpolate()

print(df)
```

## 6. Missing Value Handling Strategies

### Definition
Missing Value Handling means deciding what to do when a dataset contains missing values (`NaN`).

### Common Strategies

1. Remove rows → `dropna()`
2. Replace with a fixed value → `fillna()`
3. Replace with Mean → `fillna(df["Age"].mean())`
4. Replace with Median → `fillna(df["Age"].median())`
5. Replace with Mode → `fillna(df["City"].mode()[0])`
6. Estimate using nearby values → `interpolate()`

### Explanation

- `dropna()` → removes missing data.
- `fillna()` → replaces missing data.
- Mean → good when data has no major outliers.
- Median → useful when data contains outliers.
- Mode → commonly used for categorical data.
- `interpolate()` → useful for ordered/time-series data.

### Business Example
Before analyzing customer data, missing age, salary, or purchase information should be handled appropriately.

### AI/ML Example
Machine learning models generally need missing values to be handled before training. The strategy depends on the feature and the amount/pattern of missing data.

```python
### Python Example

import pandas as pd

df = pd.DataFrame({
    "Age": [20, 25, None, 35, 40],
    "Salary": [20000, 25000, None, 35000, 40000]
})

# Fill Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill Salary with mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df)
```
