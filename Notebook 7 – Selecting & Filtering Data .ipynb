# 1. `loc`

### Definition
Select rows and columns using **labels (row names and column names)**.

### Explanation
Uses row labels and column names to select data.

### Business Example
Retrieve the **Employee Name** and **Salary** columns for payroll reporting.

### AI/ML Example
Select input features (`X`) or target column (`y`) before training a model.

````python
### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, 30, 35],
    "Salary": [50000, 60000, 70000]
})

print(df.loc[:, "Age"])
print(df.loc[0])
print(df.loc[:, ["Name", "Salary"]])
````

# 2.`iloc`

### Definition
Select rows and columns using **integer index positions**.

### Explanation
Uses row and column index numbers instead of labels.

### Business Example
Retrieve the first 100 customer records for verification.

### AI/ML Example
Split features and target columns using their index positions.

````python
### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, 30, 35],
    "Salary": [50000, 60000, 70000]
})

# First row
print(df.iloc[0])

# First two rows
print(df.iloc[0:2])

# First and third columns
print(df.iloc[:, [0, 2]])
````

# 3.Boolean Indexing

### Definition
Filter rows based on a condition that returns **True** or **False**.


### Explanation
Returns only the rows where the condition evaluates to **True**.

### Business Example
Find employees whose salary is greater than ₹60,000.

### AI/ML Example
Remove invalid or unnecessary records before training a machine learning model.

````python
### Python Example
import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "David"],
    "Age": [25, 30, 35, 22],
    "Salary": [50000, 60000, 70000, 45000]
})

# Age greater than 28
print(df[df["Age"] > 28])

# Salary greater than or equal to 60000
print(df[df["Salary"] >= 60000])
````

# 4.Filtering Rows

### Definition
Select rows that satisfy **one or more conditions**.

### Explanation
- `&` → Both conditions must be **True** (AND).
- `|` → At least one condition must be **True** (OR).
- `~` → Reverses the condition (NOT).

### Business Example
Filter customers who are premium members and have spent more than ₹50,000.

### AI/ML Example
Filter high-quality data by removing outliers or selecting records that meet multiple criteria before training.

````python
### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "David"],
    "Age": [25, 30, 35, 22],
    "Salary": [50000, 60000, 70000, 45000]
})

# AND condition
print(df[(df["Age"] > 25) & (df["Salary"] >= 60000)])

# OR condition
print(df[(df["Age"] < 25) | (df["Salary"] >= 70000)])

# NOT condition
print(df[~(df["Age"] > 30)])
````

# 6. `query()`

### Definition
Filter rows using **SQL-like expressions**, making the code cleaner and easier to read.

### Explanation
- Uses expressions similar to SQL's `WHERE` clause.
- `and`, `or`, and `not` are used instead of `&`, `|`, and `~`.
- Makes complex filtering more readable.

### Business Example
Retrieve employees whose age is above 25 and salary is at least ₹60,000.

### AI/ML Example
Filter datasets before feature engineering or model training using readable conditions.

````python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "David"],
    "Age": [25, 30, 35, 22],
    "Salary": [50000, 60000, 70000, 45000]
})

# Single condition
print(df.query("Age > 28"))

# Multiple conditions
print(df.query("Age > 25 and Salary >= 60000"))

# OR condition
print(df.query("Age < 25 or Salary >= 70000"))
````

# 7.`isin()`

### Definition
Checks whether column values are present in a specified list.


### Explanation
- Returns rows where the column value matches any value in the given list.
- Opposite can be done using `~df["Column"].isin([...])`.

### Business Example
Find employees who work in the IT or Finance departments.

### AI/ML Example
Select only specific classes (e.g., "Cat" and "Dog") before training a classification model.

````python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "David", "Emma"],
    "Department": ["IT", "HR", "IT", "Finance", "Sales"]
})

# Check for IT and Finance departments
print(df[df["Department"].isin(["IT", "Finance"])])

# Check for HR and Sales departments
print(df[df["Department"].isin(["HR", "Sales"])])
````

# 8.`between()`

### Definition
Checks whether values fall within a specified range (inclusive by default).


### Explanation
- Returns rows where the value lies between the lower and upper limits.
- By default, both start and end values are included.

### Business Example
Find employees whose age is between 25 and 35 years.

### AI/ML Example
Filter records within a valid age or salary range before training a machine learning model.

````python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "David", "Emma"],
    "Age": [20, 25, 30, 35, 40]
})

# Age between 25 and 35
print(df[df["Age"].between(25, 35)])

# Age between 30 and 40
print(df[df["Age"].between(30, 40)])
````

# 9. Sorting Data

### Definition
Arrange rows in **ascending** or **descending** order based on one or more columns.


### Explanation
- `ascending=True` (default): Sorts from smallest to largest (A→Z).
- `ascending=False`: Sorts from largest to smallest (Z→A).
- Multiple columns can be used for sorting.

### Business Example
Sort products by sales amount to identify the highest-selling products.

### AI/ML Example
Sort prediction probabilities, feature importance scores, or model evaluation metrics for easier analysis.
```

````python

### Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "David"],
    "Age": [25, 30, 22, 35],
    "Salary": [50000, 70000, 45000, 60000]
})

# Sort by Salary (Ascending)
print(df.sort_values("Salary"))

# Sort by Salary (Descending)
print(df.sort_values("Salary", ascending=False))

# Sort by Age and Salary
print(df.sort_values(["Age", "Salary"]))
````
